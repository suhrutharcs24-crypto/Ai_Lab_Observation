### Vacuum Cleaner Agent Simulation

This simulation models a simple vacuum cleaner agent operating in two rooms (Room A and Room B). The agent's goal is to clean both rooms, starting from a user-specified initial room. It follows a basic logic: if the current room is dirty, it cleans it; otherwise, it moves to the other room.

```python
class Room:
    def __init__(self, name, is_dirty=True):
        self.name = name
        self.is_dirty = is_dirty

    def clean(self):
        self.is_dirty = False
        return f"Cleaning {self.name}."

    def __str__(self):
        status = "Dirty" if self.is_dirty else "Clean"
        return f"Room {self.name} ({status})"
```

```python
class VacuumCleanerAgent:
    def __init__(self, initial_room_name, rooms):
        if initial_room_name not in rooms:
            raise ValueError("Initial room must be 'A' or 'B'.")

        self.current_room = rooms[initial_room_name]
        self.rooms = rooms
        self.steps = []

    def _record_step(self, action):
        self.steps.append(f"Moved to {self.current_room.name}. {action}")

    def _all_rooms_clean(self):
        return all(not room.is_dirty for room in self.rooms.values())

    def run_agent(self):
        self.steps.append(f"Agent starts in {self.current_room.name}.")

        while not self._all_rooms_clean():
            if self.current_room.is_dirty:
                action = self.current_room.clean()
                self._record_step(action)
            else:
                # Move to the other room
                other_room_name = 'B' if self.current_room.name == 'A' else 'A'
                self.current_room = self.rooms[other_room_name]
                self._record_step(f"Moving to {self.current_room.name} as {other_room_name} is clean.")

        self.steps.append("All rooms are clean. Agent finished.")
        return self.steps
```

### Simulate the Agent

Run the vacuum cleaner agent by specifying its initial position (either 'A' or 'B'). The output will list all the actions taken by the agent until both rooms are clean.

```python
# Define the rooms, initially dirty
rooms_state = {
    'A': Room('A', is_dirty=True),
    'B': Room('B', is_dirty=True)
}

# Set the initial room for the vacuum cleaner agent
# You can change 'A' to 'B' to see how the agent behaves if it starts in Room B
initial_agent_position = 'A' # @param ['A', 'B']

# Create the agent and run the simulation
agent = VacuumCleanerAgent(initial_agent_position, rooms_state)
cleaning_steps = agent.run_agent()

# Display all the steps taken by the agent
for step in cleaning_steps:
    print(step)
```
