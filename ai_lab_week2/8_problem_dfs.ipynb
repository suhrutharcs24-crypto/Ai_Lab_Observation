```python
import copy

class State:
    def __init__(self, board, parent=None, move=None):
        self.board = board  # 3x3 matrix
        self.parent = parent
        self.move = move    # 'up', 'down', 'left', 'right'
        self.depth = 0
        if parent:
            self.depth = parent.depth + 1

    def __eq__(self, other):
        return self.board == other.board

    def __hash__(self):
        return hash(str(self.board))

    def find_empty_tile(self):
        for r in range(3):
            for c in range(3):
                if self.board[r][c] == 0:
                    return r, c
        return -1, -1

    def get_neighbors(self):
        neighbors = []
        r_empty, c_empty = self.find_empty_tile()

        moves = [
            ('up', -1, 0),
            ('down', 1, 0),
            ('left', 0, -1),
            ('right', 0, 1)
        ]

        for move_name, dr, dc in moves:
            new_r, new_c = r_empty + dr, c_empty + dc

            if 0 <= new_r < 3 and 0 <= new_c < 3:
                new_board = copy.deepcopy(self.board)
                new_board[r_empty][c_empty], new_board[new_r][new_c] = \
                    new_board[new_r][new_c], new_board[r_empty][c_empty]
                neighbors.append(State(new_board, self, move_name))
        return neighbors

def print_board(board):
    for row in board:
        print(" ".join(map(str, row)))

def print_path(state):
    path = []
    while state:
        path.append(state)
        state = state.parent

    path.reverse()
    for i, s in enumerate(path):
        print(f"--- Step {i} (Move: {s.move if s.move else 'Initial'}) ---")
        print_board(s.board)
        print()

def dfs(initial_state_matrix, goal_state_matrix):
    initial_state = State(initial_state_matrix)
    goal_state = State(goal_state_matrix)

    stack = [initial_state]
    visited = set()
    visited.add(initial_state)

    while stack:
        current_state = stack.pop()

        if current_state == goal_state:
            print("Goal state reached!")
            print_path(current_state)
            return True

        for neighbor in current_state.get_neighbors():
            if neighbor not in visited:
                visited.add(neighbor)
                stack.append(neighbor)

    print("Goal state not reachable.")
    return False

# --- Example Usage ---
# Define initial and goal states
# 0 represents the empty tile
initial_state = [
    [1, 2, 3],
    [4, 0, 5],
    [6, 7, 8]
]

goal_state = [
    [1, 2, 3],
    [4, 5, 0],
    [6, 7, 8]
]

print("Initial State:")
print_board(initial_state)
print("\nGoal State:")
print_board(goal_state)
print("\nSearching for solution using DFS...")

dfs(initial_state, goal_state)
```
