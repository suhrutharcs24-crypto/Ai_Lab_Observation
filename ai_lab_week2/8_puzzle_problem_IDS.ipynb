## 8-Puzzle Problem using Iterative Deepening Search (IDS)

The 8-puzzle is a sliding puzzle that consists of a 3x3 grid with 8 numbered tiles and one blank space. The goal is to rearrange the tiles from an initial configuration to a goal configuration by sliding tiles into the blank space.

Iterative Deepening Search (IDS) combines the benefits of Breadth-First Search (completeness and optimality for uniform cost) and Depth-First Search (memory efficiency). It performs a series of depth-limited searches with increasing depth limits.

```python
import collections

def get_neighbors(state):
    """Generates all possible next states from the current state."""
    rows = 3
    cols = 3
    # Find the position of the blank space (0)
    zero_index = state.index(0)
    zero_row, zero_col = zero_index // cols, zero_index % cols

    neighbors = []
    # Possible moves: (dr, dc) for (up, down, left, right)
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row, new_col = zero_row + dr, zero_col + dc

        # Check if the new position is within the grid boundaries
        if 0 <= new_row < rows and 0 <= new_col < cols:
            new_state_list = list(state)
            # Calculate the index of the tile to swap with the blank space
            swap_index = new_row * cols + new_col

            # Swap the blank space with the tile at the new position
            new_state_list[zero_index], new_state_list[swap_index] = \
                new_state_list[swap_index], new_state_list[zero_index]

            neighbors.append(tuple(new_state_list))
    return neighbors

def print_puzzle(state):
    """Prints the 8-puzzle state in a 3x3 grid format."""
    for i in range(3):
        print(state[i*3 : (i+1)*3])
    print("\n")
```

### Depth-Limited Search (DLS)

This is a helper function for IDS. It explores as deeply as possible into the search tree but only up to a specified `limit`.

```python
def dls(initial_state, goal_state, limit):
    """Performs Depth-Limited Search (DLS) from the initial state to the goal state."""
    # A stack to store (state, path, depth) tuples
    stack = collections.deque([(initial_state, [initial_state], 0)])
    visited = set()

    while stack:
        current_state, path, depth = stack.pop()

        if current_state == goal_state:
            return path  # Goal found

        if depth < limit:
            for neighbor in get_neighbors(current_state):
                if neighbor not in visited:
                    visited.add(neighbor)
                    stack.append((neighbor, path + [neighbor], depth + 1))

    return None  # Goal not found within the limit
```

### Iterative Deepening Search (IDS)

IDS calls DLS repeatedly with increasing depth limits until the goal is found or the maximum allowed depth is reached.

```python
def ids(initial_state, goal_state, max_depth):
    """Performs Iterative Deepening Search (IDS) to find the goal state."""
    for depth in range(max_depth + 1):
        print(f"Searching with depth limit: {depth}")
        result = dls(initial_state, goal_state, depth)
        if result is not None:
            return result
    return None  # Goal not found within max_depth
```

### User Input and Execution

Now, let's define the initial and goal states, and the maximum search depth, then run the IDS algorithm.

```python
def get_user_puzzle_input(prompt):
    while True:
        try:
            user_input = input(f"{prompt} (e.g., '1 2 3 4 5 6 7 8 0' for 0 as blank): ")
            state = tuple(map(int, user_input.split()))
            if len(state) == 9 and all(0 <= x <= 8 for x in state) and len(set(state)) == 9:
                return state
            else:
                print("Invalid input. Please enter 9 unique numbers from 0 to 8, separated by spaces.")
        except ValueError:
            print("Invalid input. Please enter numbers separated by spaces.")

def get_user_depth_input(prompt):
    while True:
        try:
            user_input = int(input(f"{prompt}: "))
            if user_input >= 0:
                return user_input
            else:
                print("Depth must be a non-negative integer.")
        except ValueError:
            print("Invalid input. Please enter an integer.")

# Get initial state from user
initial_state = get_user_puzzle_input("Enter the initial state of the 8-puzzle")

# Get goal state from user
goal_state = get_user_puzzle_input("Enter the goal state of the 8-puzzle")

# Get maximum depth from user
max_depth = get_user_depth_input("Enter the maximum depth for the search traversal")

print("\nInitial State:")
print_puzzle(initial_state)

print("Goal State:")
print_puzzle(goal_state)

print(f"Max Search Depth: {max_depth}\n")

# Run IDS
path = ids(initial_state, goal_state, max_depth)

if path:
    print(f"Goal reached in {len(path) - 1} steps. Path:\n")
    for step, state in enumerate(path):
        print(f"Step {step}:")
        print_puzzle(state)
else:
    print("Goal not found within the specified maximum depth.")
```

```python
# Re-run IDS with a new maximum depth

new_max_depth = get_user_depth_input("Enter a new maximum depth for the search traversal")

print(f"\nRe-running search with new Max Search Depth: {new_max_depth}\n")

# Run IDS with updated max_depth
path = ids(initial_state, goal_state, new_max_depth)

if path:
    print(f"Goal reached in {len(path) - 1} steps. Path:\n")
    for step, state in enumerate(path):
        print(f"Step {step}:")
        print_puzzle(state)
else:
    print("Goal not found within the specified new maximum depth.")
```
