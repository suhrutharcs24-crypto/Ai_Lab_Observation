# 4-Queens Problem using Hill Climbing
This notebook implements the Hill Climbing algorithm to solve the 4-queens problem.

```python
import random

def get_conflicts(state):
    """Calculate the number of mutual attacks between queens."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbors(state):
    """Generate all possible states by moving one queen within its column."""
    neighbors = []
    n = len(state)
    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbors.append(tuple(neighbor))
    return neighbors

def hill_climbing(n=4, initial_state=None):
    """Standard Hill Climbing algorithm with an optional initial state."""
    if initial_state is not None:
        current_state = tuple(initial_state)
    else:
        current_state = tuple(random.randint(0, n - 1) for _ in range(n))

    current_conflicts = get_conflicts(current_state)

    while True:
        neighbors = get_neighbors(current_state)
        if not neighbors:
            break

        # Find the neighbor(s) with the minimum conflicts
        neighbor_conflicts = [(neighbor, get_conflicts(neighbor)) for neighbor in neighbors]
        best_neighbor, min_conflicts = min(neighbor_conflicts, key=lambda x: x[1])

        # If we cannot find a strictly better neighbor, we stop (reached local/global minimum)
        if min_conflicts >= current_conflicts:
            break

        current_state = best_neighbor
        current_conflicts = min_conflicts

    return current_state, current_conflicts

def print_board(state):
    """Helper function to print the board nicely."""
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

def solve_4_queens(initial_state=None):
    """Solve 4-queens using Hill Climbing with a customizable starting position."""
    if initial_state is not None:
        print(f"Starting hill climbing with initial state: {initial_state}")
        state, conflicts = hill_climbing(4, initial_state=initial_state)
        if conflicts == 0:
            print("Solution found successfully from your initial position!")
            print_board(state)
        else:
            print(f"Stuck in a local minimum at state {state} with {conflicts} conflicts.")
            print_board(state)
            print("\n--- Restarting with random positions to find a solution --- ")

    if initial_state is None or conflicts > 0:
        attempts = 1 if initial_state is None else 2
        while True:
            state, conflicts = hill_climbing(4)
            if conflicts == 0:
                print(f"Solution found in {attempts} attempt(s)!")
                print(f"State: {state}")
                print_board(state)
                break
            attempts += 1

# Try solving with a custom initial position (e.g. all queens on the first row: [0, 0, 0, 0])
user_initial_state = [3, 1, 2, 0]
solve_4_queens(initial_state=user_initial_state)
```
